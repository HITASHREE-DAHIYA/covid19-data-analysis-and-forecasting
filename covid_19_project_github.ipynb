Growth Rate Analysis

```python
df['Daily growth %']= df['Confirmed'].pct_change()*100
fig = px.line(df, x='Date', y='Daily growth %', title= 'Daily Growth Rate')
fig.show()
```

Prepairing Data For prediction

```python
Prophet_df=df[['Date','Confirmed']]
Prophet_df.columns = ['ds','y']
```

Training the Model

```python
model =Prophet()
model.fit(Prophet_df)
```

Future predictions for the week 

```python
future= model.make_future_dataframe(periods=7)
forecast= model.predict(future)
print(forecast[['ds','yhat', 'yhat_lower', 'yhat_upper']].tail(7))
```

visualization of Prediction

```python
fig1=model.plot(forecast)
fig2= model.plot_components(forecast)
```

```python
fig= px.line(forecast, x='ds', y='yhat',title='predicted COVID cases for next 7 days')
fig.show()
```
